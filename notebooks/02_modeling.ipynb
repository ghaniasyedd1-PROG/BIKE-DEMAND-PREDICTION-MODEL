# 02 — Modelling and Comparison

This notebook reviews the regression and classification models kept in the project. Regression and classification model code is separated into `src/models/regression/` and `src/models/classification/`. The training workflow uses chronological train/test data, tunes model settings with `TimeSeriesSplit`, and reports results on the latest held-out records. ANN and KNN have been removed.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image
ROOT = Path.cwd() if (Path.cwd() / "models/metrics.json").exists() else Path.cwd().parent
metrics = json.loads((ROOT / "models/metrics.json").read_text())
comparison = json.loads((ROOT / "models/model_comparison.json").read_text())
```

## Time split, target creation and leakage prevention

```python
train = pd.read_csv(ROOT / "data/processed/train.csv", parse_dates=["datetime"])
test = pd.read_csv(ROOT / "data/processed/test.csv", parse_dates=["datetime"])
print("Train:", train.shape, train.datetime.min(), "to", train.datetime.max())
print("Test:", test.shape, test.datetime.min(), "to", test.datetime.max())
print("Target cutoffs calculated on training rows only:", metrics["classification_cutoffs_train_only"])
print("casual/registered are not in model features.")
```

## Regression comparison

**Pipeline steps:** preprocess features, fit the scikit-learn estimator, tune the listed hyperparameters on chronological training folds, evaluate on the held-out latest period, and display a model-specific metric plot below.

```python
reg_table = pd.DataFrame(comparison["regression"]).sort_values("R2", ascending=False)
display(reg_table)
reg_table.plot(x="model", y="R2", kind="bar", legend=False, title="Regression R² comparison", rot=35, figsize=(9,4))
plt.tight_layout()
plt.show()
```

## Classification comparison

**Pipeline steps:** preprocess features, fit the scikit-learn estimator, tune the listed hyperparameters on chronological training folds, evaluate on the held-out latest period, and display a model-specific metric plot below.

```python
clf_table = pd.DataFrame(comparison["classification"]).sort_values("macro_f1", ascending=False)
display(clf_table)
clf_table.plot(x="model", y="accuracy", kind="bar", legend=False, title="Classification accuracy comparison", rot=35, figsize=(9,4))
plt.tight_layout()
plt.show()
```

## Decision Tree — Regression

Learns decision rules by splitting feature values; depth and leaf size are tuned to balance fit and generalisation.

**Training:** the model is fitted in a scikit-learn Pipeline with the ColumnTransformer. **Tuning:** GridSearchCV with TimeSeriesSplit is used where a search grid is defined; see the best parameters below. Naive Bayes uses its standard GaussianNB settings; the other required models use GridSearchCV with TimeSeriesSplit.

**Pipeline steps:** preprocess features, fit the scikit-learn estimator, tune the listed hyperparameters on chronological training folds, evaluate on the held-out latest period, and display a model-specific metric plot below.

```python
result = metrics["regression"]["models"].get("decision_tree")
print("Held-out regression result for decision_tree:")
print(result)
if result:
    score_name = "R2"
    plt.figure(figsize=(4, 2.5))
    plt.bar(["decision_tree"], [result[score_name]])
    plt.ylim(0, 1)
    plt.ylabel(score_name)
    plt.title("Decision Tree — Regression")
    plt.tight_layout()
    plt.show()
```
